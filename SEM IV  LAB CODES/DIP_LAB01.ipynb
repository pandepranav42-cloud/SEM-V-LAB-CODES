```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('/content/LAB_01.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

print("image Shape : ", img.shape)

# Display Image
plt.imshow(img)
plt.title("Acquired Image ")
plt.axis('off')
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

plt.figure(figsize=(6, 6))
plt.imshow(gray, cmap='gray')
plt.title("Image Divided into Grid (Pixel Representation)")

step = 50

for i in range(0, gray.shape[0], step):
    plt.axhline(i, color='red', linewidth=0.6)

for j in range(0, gray.shape[1], step):
    plt.axvline(j, color='green', linewidth=0.6)

plt.axis('off')
plt.show()
```

```python
print("pixel intensity values (Sample):\n")

print(gray[0:10 ,0:10])
```

```python
sampled = cv2.resize(gray, (20,20))

plt.figure(figsize=(10, 4))

plt.subplot(1,2,1)
plt.imshow(gray, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(sampled, cmap='gray')
plt.title("Sampled Image (Reduced Resolution)")
plt.axis('off')

plt.show()
```

```python
import numpy as np

levels = 5

quantized = np.floor(gray / (256/levels))* (256/levels)
quantized = quantized.astype(np.uint8)

plt.figure(figsize = (10,4))

plt.subplot(1,2,1)
plt.imshow(gray , cmap='gray')
plt.title("Orignal Image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(gray , cmap= 'gray')
plt.title("Quantized Image (16 Gray Levels )")
plt.axis('off')

plt.show()
```

```python
print("Original image pixel intensity values (Sample 10x10):\n")
print(img[0:10 ,0:10])

print("\nQuantized image pixel intensity values (Sample 10x10):\n")
print(quantized[0:10 ,0:10])
```
