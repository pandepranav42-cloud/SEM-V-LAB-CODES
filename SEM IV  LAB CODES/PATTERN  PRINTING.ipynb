### Pattern Printing Examples

Here are some common pattern printing codes in Python using nested loops.

```python
# 1. Solid Rectangle Pattern
rows = 5
cols = 5
for i in range(rows):
    for j in range(cols):
        print('*', end=' ')
    print()
```

```python
# 2. Hollow Rectangle Pattern
rows = 5
cols = 5
for i in range(rows):
    for j in range(cols):
        if i == 0 or i == rows - 1 or j == 0 or j == cols - 1:
            print('*', end=' ')
        else:
            print(' ', end=' ')
    print()
```

```python
# 3. Right Half Pyramid
rows = 5
for i in range(1, rows + 1):
    for j in range(1, i + 1):
        print('*', end=' ')
    print()
```

```python
# 4. Inverted Right Half Pyramid
rows = 5
for i in range(rows, 0, -1):
    for j in range(1, i + 1):
        print('*', end=' ')
    print()
```

```python
# 5. Left Half Pyramid
rows = 5
for i in range(1, rows + 1):
    for j in range(1, rows - i + 1):
        print(' ', end=' ')
    for j in range(1, i + 1):
        print('*', end=' ')
    print()
```

### More Pattern Printing Examples

```python
# 6. Full Pyramid (Equilateral Triangle)
rows = 5
for i in range(1, rows + 1):
    for j in range(1, rows - i + 1):
        print(' ', end='')
    for k in range(1, 2 * i):
        print('*', end='')
    print()
```

```python
# 7. Inverted Full Pyramid
rows = 5
for i in range(rows, 0, -1):
    for j in range(1, rows - i + 1):
        print(' ', end='')
    for k in range(1, 2 * i):
        print('*', end='')
    print()
```

```python
# 8. Diamond Pattern
rows = 5

# Upper part
for i in range(1, rows + 1):
    for j in range(1, rows - i + 1):
        print(' ', end='')
    for k in range(1, 2 * i):
        print('*', end='')
    print()

# Lower part
for i in range(rows - 1, 0, -1):
    for j in range(1, rows - i + 1):
        print(' ', end='')
    for k in range(1, 2 * i):
        print('*', end='')
    print()
```

```python
# 9. Number Pyramid
rows = 5
for i in range(1, rows + 1):
    for j in range(1, rows - i + 1):
        print(' ', end=' ')
    for k in range(1, i + 1):
        print(k, end=' ')
    for l in range(i - 1, 0, -1):
        print(l, end=' ')
    print()
```

```python
# 10. Alphabet Pyramid
rows = 5
for i in range(1, rows + 1):
    for j in range(1, rows - i + 1):
        print(' ', end=' ')
    for k in range(1, i + 1):
        print(chr(64 + k), end=' ')
    for l in range(i - 1, 0, -1):
        print(chr(64 + l), end=' ')
    print()
```

```python
# 11. Increasing Number Triangle
rows = 5
current_num = 1
for i in range(1, rows + 1):
    for j in range(1, i + 1):
        print(current_num, end=' ')
        current_num += 1
    print()
```

```python
# 12. Decreasing Number Triangle
rows = 5
for i in range(rows, 0, -1):
    for j in range(1, i + 1):
        print(j, end=' ')
    print()
```

```python
# 13. Alphabet Triangle (Increasing Letter)
rows = 5
for i in range(1, rows + 1):
    for j in range(1, i + 1):
        print(chr(64 + j), end=' ')
    print()
```

```python
# 14. Alphabet Triangle (Repeated Letter)
rows = 5
for i in range(1, rows + 1):
    for j in range(1, i + 1):
        print(chr(64 + i), end=' ')
    print()
```

```python
# 15. Binary Triangle
rows = 5
for i in range(1, rows + 1):
    for j in range(1, i + 1):
        if (i + j) % 2 == 0:
            print('1', end=' ')
        else:
            print('0', end=' ')
    print()
```

```python
# 16. Hourglass Pattern
rows = 5

# Upper part
for i in range(rows, 0, -1):
    for j in range(1, rows - i + 1):
        print(' ', end='')
    for k in range(1, 2 * i):
        print('*', end='')
    print()

# Lower part (excluding the middle row if rows is odd)
for i in range(2, rows + 1):
    for j in range(1, rows - i + 1):
        print(' ', end='')
    for k in range(1, 2 * i):
        print('*', end='')
    print()
```

```python
# 17. Right Pascal Triangle
rows = 5

# Upper part
for i in range(1, rows + 1):
    for j in range(1, i + 1):
        print('*', end=' ')
    print()

# Lower part
for i in range(rows - 1, 0, -1):
    for j in range(1, i + 1):
        print('*', end=' ')
    print()
```

```python
# 18. Left Pascal Triangle
rows = 5

# Upper part
for i in range(1, rows + 1):
    for j in range(1, rows - i + 1):
        print(' ', end=' ')
    for k in range(1, i + 1):
        print('*', end=' ')
    print()

# Lower part
for i in range(rows - 1, 0, -1):
    for j in range(1, rows - i + 1):
        print(' ', end=' ')
    for k in range(1, i + 1):
        print('*', end=' ')
    print()
```

```python
# 38. X Pattern (using spaces for hollow parts)
rows = 5 # Should be odd
for i in range(1, rows + 1):
    for j in range(1, rows + 1):
        if i == j or i + j == rows + 1:
            print('*', end=' ')
        else:
            print(' ', end=' ')
    print()
```

```python
# 20. Hollow Rhombus
rows = 5
for i in range(1, rows + 1):
    for j in range(1, rows - i + 1):
        print(' ', end='')
    for k in range(1, rows + 1):
        if i == 1 or i == rows or k == 1 or k == rows:
            print('*', end='')
        else:
            print(' ', end='')
    print()
```

```python
# 21. Palindromic Number Pyramid
rows = 5
for i in range(1, rows + 1):
    for j in range(1, rows - i + 1):
        print(' ', end=' ')
    for k in range(i, 0, -1):
        print(k, end=' ')
    for l in range(2, i + 1):
        print(l, end=' ')
    print()
```

```python
# 41. Swastika Pattern
# This pattern is complex and typically requires a specific logic for each segment.
# A common approach is to use a matrix or conditional printing for each cell.
rows = 7 # Example for a 7x7 swastika, typically odd size
size = rows
for i in range(size):
    for j in range(size):
        if i == size // 2 or j == size // 2 or \
           (i == 0 and j >= size // 2) or (j == 0 and i <= size // 2) or \
           (i == size - 1 and j <= size // 2) or (j == size - 1 and i >= size // 2):
            print('*', end=' ')
        else:
            print(' ', end=' ')
    print()
```

```python
# 42. Plus Sign Pattern
rows = 5 # Should be odd
for i in range(rows):
    for j in range(rows):
        if i == rows // 2 or j == rows // 2:
            print('*', end=' ')
        else:
            print(' ', end=' ')
    print()
```

```python
# 28. Inverted Number Right Triangle
rows = 5
for i in range(rows, 0, -1):
    for j in range(1, i + 1):
        print(j, end=' ')
    print()
```

```python
# 29. Right Aligned Number Triangle
rows = 5
for i in range(1, rows + 1):
    for j in range(1, rows - i + 1):
        print(' ', end=' ')
    for k in range(1, i + 1):
        print(k, end=' ')
    print()
```

```python
# 30. Left Aligned Inverted Number Triangle
rows = 5
for i in range(rows, 0, -1):
    for j in range(1, i + 1):
        print(j, end=' ')
    print()
```

```python
# 31. Right Triangle Star Pattern (Descending on one side)
rows = 5
for i in range(1, rows + 1):
    for j in range(i, rows + 1):
        print('*', end=' ')
    print()
```

```python
# 32. Inverted Left Half Pyramid
rows = 5
for i in range(rows, 0, -1):
    for j in range(1, rows - i + 1):
        print(' ', end=' ')
    for k in range(1, i + 1):
        print('*', end=' ')
    print()
```

```python
# 33. Downward Half Pyramid (Numbers)
rows = 5
for i in range(rows, 0, -1):
    for j in range(1, i + 1):
        print(i, end=' ')
    print()
```

```python
# 34. Upward Half Pyramid (Numbers)
rows = 5
for i in range(1, rows + 1):
    for j in range(1, i + 1):
        print(i, end=' ')
    print()
```

```python
# 35. Snake Pattern (alternating characters)
rows = 5
cols = 5
for i in range(rows):
    if i % 2 == 0:
        for j in range(cols):
            print('*', end=' ')
    else:
        for j in range(cols):
            print('#', end=' ')
    print()
```
