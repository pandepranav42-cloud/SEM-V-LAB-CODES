```python
import numpy as np
import pandas as pd

df = pd.read_csv("student_performance_1000.csv")
df.head()
```

```python
print("Dataset Shape:", df.shape)
print("\nDataset Info:\n")
df.info()
```

```python
df.describe()
```

```python
### Rules:
# Object / String → Qualitative (Categorical)
# Integer → Quantitative Discrete
# Float → Quantitative Continuous


def classify_data_type(series):   #  defined function
    if series.dtype == 'object':
        return "Qualitative"
    elif np.issubdtype(series.dtype, np.integer):
        return "Quantitative - Discrete"
    elif np.issubdtype(series.dtype, np.floating):
        return "Quantitative - Continuous"
    else:
        return "Unknown"
```

```python
## 📏 Step 5: Logic for Scale of Measurement

### Scales:
# Nominal → Names / categories (no order)
# Ordinal → Ordered categories (Grades)
# Interval → Equal intervals, no true zero
# Ratio → True zero exists
```

```python
def identify_scale(column):
    col = column.lower()
    
    if col in ["student_name", "grade"]:
        if col == "grade":
            return "Ordinal"
        return "Nominal"
    
    if col in ["s1", "s2", "s3", "s4", "s5"]:
        return "Interval"
    
    if col in ["study_hours_per_week", "sgpa", "sr_no"]:
        return "Ratio"
    
    return "Ratio (Assumed)"
```

```python
results = []

for col in df.columns:
    data_type = classify_data_type(df[col])
    scale = identify_scale(col)
    
    results.append([col, 
                    
                    data_type, scale])

classification_df = pd.DataFrame(
    results,
    columns=["Attribute", "Data Type", "Scale of Measurement"]
)

classification_df
```

```python
## ✅ Step 7: Final Interpretation

### Observations:
# Student_Name → Qualitative, Nominal
# Study_Hours → Quantitative, Discrete, Ratio
# Subject Marks → Quantitative, Continuous, Interval
# SGPA → Quantitative, Continuous, Ratio
# Grade → Qualitative, Ordinal
```

```python
classification_df.to_csv(
    "student_data_classification_result.csv",
    index=False
)

print("Classification report saved successfully")
```
