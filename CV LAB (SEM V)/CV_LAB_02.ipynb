```python
import cv2
import matplotlib.pyplot as plt

# Read original image
img = cv2.imread("/content/image.jpg")

# Convert BGR → RGB
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# BGR is the original format read by OpenCV
bgr = img

# Convert BGR → Grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Convert BGR → HSV
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# Display all images
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(rgb)
plt.title("Original / RGB")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB))
plt.title("BGR")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(hsv)
plt.title("HSV")
plt.axis("off")

plt.tight_layout()
plt.show()
```
