```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("image.jpg")
```

```python
bgr = img
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
grayscale = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
ycbcr = cv2.cvtColor(img, cv2.COLOR_BGR2YCrCb)
```

```python
# display
plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(bgr)
plt.title("BGR")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(rgb)
plt.title("RGB")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(grayscale, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(hsv)
plt.title("HSV")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(ycbcr)
plt.title("YCbCr")
plt.axis("off")

plt.show()
```
