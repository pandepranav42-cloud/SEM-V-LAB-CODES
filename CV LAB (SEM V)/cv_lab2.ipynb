```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
image_path = '/content/photo.jpg'
image = cv2.imread(image_path)

if image is None:
    print("Error loading image")
else:
    print("Image loaded successfully")
```

```python
plt.figure(figsize=(16, 8))
plt.subplot(1,4,1)
plt.imshow(image)
plt.title("BGR")
plt.axis('off')

plt.subplot(1,4,2)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("RGB")
plt.axis('off')

plt.subplot(1,4,3)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2GRAY), cmap="gray")
plt.title("GRAY")
plt.axis('off')

plt.subplot(1,4,4)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2HSV))
plt.title("HSV")
plt.axis('off')

plt.show()
```
