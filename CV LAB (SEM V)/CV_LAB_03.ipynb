```python
import cv2
import matplotlib.pyplot as plt

# Read image
img = cv2.imread("/content/img.jpg")

# Convert BGR to RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Apply Gaussian filter
gaussian = cv2.GaussianBlur(img_rgb, (5, 5), 0)

# Display using subplot
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gaussian)
plt.title("Gaussian Filter")
plt.axis("off")

plt.tight_layout()
plt.show()
```
