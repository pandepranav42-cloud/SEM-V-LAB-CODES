```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
# load image
image_path = "/content/photo.jpg"
image = cv2.imread(image_path)

if image is None:
  print("Error loading image.")
else:
  print("Image loaded successfully.")
```

```python
color_img = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
# Gaussian filter
gauss_blur = cv2.GaussianBlur(color_img, (7, 7), 0)
```

```python
# output
plt.figure(figsize=(16, 8))

plt.subplot(1, 2, 1)
plt.title("Original Image")
plt.imshow(color_img)
plt.axis("off")

plt.subplot(1, 2, 2)
plt.title("Gaussian Blur")
plt.imshow(gauss_blur)
plt.axis("off")

plt.show()
```
