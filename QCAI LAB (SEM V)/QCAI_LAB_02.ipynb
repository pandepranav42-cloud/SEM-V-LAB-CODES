```python
import numpy as np
```

```python
# Identity gate
I = np.array([
    [1, 0],
    [0, 1]
])

# Pauli-X gate (NOT gate)
X = np.array([
    [0, 1],
    [1, 0]
])

# Pauli-Y gate
Y = np.array([
    [0, -1j],
    [1j, 0]
])

# Pauli-Z gate
Z = np.array([
    [1, 0],
    [0, -1]
])

# Hadamard gate
H = (1 / np.sqrt(2)) * np.array([
    [1, 1],
    [1, -1]
])
```

```python
# |0> state
zero = np.array([1, 0])

# |1> state
one = np.array([0, 1])
```

```python
print("X|0> =", X @ zero)
print("Y|0> =", Y @ zero)
print("Z|0> =", Z @ zero)
print("H|0> =", H @ zero)
```

```python
state = H @ zero

print("Quantum state:", state)

probability_0 = abs(state[0])**2
probability_1 = abs(state[1])**2

print("Probability of |0> =", probability_0)
print("Probability of |1> =", probability_1)
```

```python
print("X^2 = I :", np.allclose(X @ X, I))
print("Y² = I :", np.allclose(Y @ Y, I))
print("Z² = I :", np.allclose(Z @ Z, I))
print("H² = I :", np.allclose(H @ H, I))
```

```python
# State after applying H to |0>
state = H @ zero

# Simulate 1000 measurements
results = np.random.choice(
    [0, 1],
    size=1000,
    p=[abs(state[0])**2, abs(state[1])**2]
)

print("Number of 0 results:", np.sum(results == 0))
print("Number of 1 results:", np.sum(results == 1))
```
