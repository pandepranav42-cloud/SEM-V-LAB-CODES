```python
from qiskit.quantum_info import Statevector
```

```python
#Circuit 1 : |0> ->H -> Superposition
qc0 = QuantumCircuit(1)
#Initial state is |0>
qc0.h(0)
print("Circuit for |0>followed by hadamard;")
print(qc0)
```

```python
#Get Final Statevector
state0=Statevector. from_instruction(qc0)
print("\nFinal state:")
print(state0)
print("\nProbabilities:")
print(state0.probabilities_dict())
```

```python
from qiskit import QuantumCircuit

# Circuit 2: |1> -> H -> Superposition

qc1 = QuantumCircuit(1)

# Prepare |1> state
qc1.x(0)

# Apply Hadamard gate
qc1.h(0)

print("\nCircuit for |1> followed by Hadamard:")
print(qc1)
```

```python
from qiskit.quantum_info import Statevector

# Get Final Statevector
state1 = Statevector.from_instruction(qc1)

print("\nFinal state:")
print(state1)

print("\nProbabilities:")
print(state1.probabilities_dict())
```
